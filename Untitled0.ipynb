```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual styling for our charts later
sns.set_theme(style="whitegrid")
np.random.seed(42) # Ensures our random data stays consistent every time we run it

# --- 1. GENERATE SAMPLE RETAIL / FMCG INVENTORY DATA ---
skus = [
    "SKU-101-MaizeMeal",
    "SKU-102-CookingOil",
    "SKU-103-LongLifeMilk",
    "SKU-104-CannedBeans",
    "SKU-105-LaundryDetergent"
]

n_records = 300
data = {
    "Transaction_Date": pd.date_range(start="2026-01-01", periods=n_records, freq="D"),
    "SKU": np.random.choice(skus, n_records),
    "Units_Sold": np.random.poisson(lam=45, size=n_records), # Simulating daily sales volume
    "Lead_Time_Days": np.random.choice([3, 5, 7, 10], n_records, p=[0.4, 0.3, 0.2, 0.1]) # Supplier delivery lead times in days
}

df = pd.DataFrame(data)

print("--- Sample Inventory Transactions (First 5 Rows) ---")
print(df.head())
```

```python
# --- 2. CALCULATE SKU METRICS & INVENTORY POLICY ---
inventory_summary = df.groupby("SKU").agg(
    Avg_Daily_Demand=("Units_Sold", "mean"),
    Demand_Std_Dev=("Units_Sold", "std"),
    Avg_Lead_Time=("Lead_Time_Days", "mean")
).reset_index()

# 95% Service Level Z-score
Z_SCORE = 1.65

# Safety Stock calculation
inventory_summary["Safety_Stock"] = (
    Z_SCORE * inventory_summary["Demand_Std_Dev"] * np.sqrt(inventory_summary["Avg_Lead_Time"])
).round(0)

# Reorder Point (ROP) calculation
inventory_summary["Reorder_Point"] = (
    (inventory_summary["Avg_Daily_Demand"] * inventory_summary["Avg_Lead_Time"]) + inventory_summary["Safety_Stock"]
).round(0)

print("\n--- Calculated Inventory Optimization Policy ---")
print(inventory_summary.to_string(index=False))
```

```python
# --- 3. DATA VISUALIZATION ---
plt.figure(figsize=(10, 5))
barplot = sns.barplot(
    data=inventory_summary,
    x="SKU",
    y="Reorder_Point",
    palette="viridis",
    hue="SKU",
    legend=False
)

# Overlay safety stock values as reference text inside the bars
for index, row in inventory_summary.iterrows():
    plt.text(index, row["Reorder_Point"] / 2, f"Safety Stock:\n{int(row['Safety_Stock'])} units",
             color="white", ha="center", fontweight="bold", fontsize=9)

plt.title("Calculated Inventory Reorder Points (ROP) per SKU", fontsize=14, fontweight="bold")
plt.xlabel("Product SKU", fontsize=12)
plt.ylabel("Reorder Point (Units)", fontsize=12)
plt.xticks(rotation=15)
plt.tight_layout()

plt.savefig("inventory_reorder_analysis.png")
plt.show()
```
